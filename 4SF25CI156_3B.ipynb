```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
#read the dataset as pandas dataframe and display first 5 rows
df=pd.read_csv('amazon.csv')
df.head(5)
```

```python
# Display the number of rows and columns
print("Dataset Shape:\n", df.shape,"\n")
```

```python
# Display all column names
print("Column Names:\n",df.columns,"\n")
```

```python
# Display the data type of each column
print("Data Types:\n",df.dtypes,"\n")
```

```python
#Describe the DataFrame
print("Statistical summary:\n",df.describe())
```

```python
#The following columns are removed and remaining colums are printed
column_drop = ['product_id','about_product','user_id','user_name','review_id','review_title','review_content','img_link','product_link']
df.drop(columns=column_drop, inplace=True)

print("Column Names:\n",df.columns,"\n")
display(df.head())
```

```python
#this removes the rupee symbol and commas from discounted_price and actual_price column and then convert to float
df['discounted_price']=(df['discounted_price'].str.replace('₹','').str.replace(',','').astype(float))
df['actual_price']=(df['actual_price'].str.replace('₹','').str.replace(',', '').astype(float))

#this removes the percentage sign and convert the discount column to float
df['discount_percentage']=(df['discount_percentage'].str.replace('%','').astype(float))

#this removes commas from rating_count and convert it to a numeric column
df['rating_count']=df['rating_count'].str.replace(',', '')
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce')


#this converts rating to numeric datatype
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

print("Data types after transformation:")
print(df.dtypes)
```

```python
#To find the missing values in each columns
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
#Replaceing the missing numerical values with their column means
df['rating'] = df['rating'].fillna(df['rating'].mean())
df['rating_count'] = df['rating_count'].fillna(df['rating_count'].mean())

print("Missing values after imputation:")
print(df.isnull().sum())
```

```python
#Creating two new columns price_savings and main_category
# price_savings calculates the amount saved after applying the discount
df['price_savings'] = df['actual_price'] - df['discounted_price']

# Extract the main category from the category column 
df['main_category'] = df['category'].str.split('|').str[0]

display(df[['product_name','actual_price','discounted_price','price_savings','category','main_category']].head())
```

```python
print("Final Dataset Shape:", df.shape)

print("\nFinal Data Types:\n", df.dtypes)


print("\nMissing Values:\n",df.isnull().sum())


print("\nFinal Dataset:")
display(df.head())
```

```python
#Number of Products in Each Main Category
category_counts=df['main_category'].value_counts()

print("Number of Main Categories:", df['main_category'].nunique())
print("\nProducts in Each Category:")
print(category_counts)
```

```python
#Average Product Rating by Category
category_rating=df.groupby('main_category')['rating'].mean()

print("Average Rating by Category:")
print(category_rating.sort_values(ascending=False))
```

```python
#Top 10 most-rated products
top_products=df.sort_values(by='rating_count', ascending=False)

print("Top 10 Most-Rated Products:")
display(top_products[['product_name', 'rating', 'rating_count']].head(10))
```

```python
#products with highest discount
top_discounted=df.sort_values(by='discount_percentage',ascending=False)

print("Top 10 Products with Highest Discounts:")
display(top_discounted[['product_name', 'actual_price', 'discounted_price','discount_percentage']].head(10))
```

```python
print("Overall Dataset Insights")

print("\nAverage Product Rating:", round(df['rating'].mean(), 2))
print("Average Discount:", round(df['discount_percentage'].mean(), 2), "%")
print("Average Discounted Price:", round(df['discounted_price'].mean(), 2))
print("Average Actual Price:", round(df['actual_price'].mean(), 2))
print("Average Price Savings:", round(df['price_savings'].mean(), 2))
```

```python
# Bar Plot: Number of products in each main category

categories = df['main_category'].value_counts()
plt.figure(figsize=(10, 5))
plt.bar(categories.index, categories.values, color='orange')
plt.xlabel('Product Categories')
plt.ylabel('Number of Products')
plt.title('Number of Products in Each Category')
plt.xticks(rotation=90)
plt.show()
```

```python
# Bar Plot: Average rating in each main category

avg_rating = df.groupby('main_category')['rating'].mean()
plt.figure(figsize=(10, 5))
plt.bar(avg_rating.index, avg_rating.values, color='skyblue')
plt.xlabel('Product Categories')
plt.ylabel('Average Rating')
plt.title('Average Product Rating by Category')
plt.xticks(rotation=90)
plt.show()
```

```python
# Bar Plot: Average discount in each main category

avg_discount = df.groupby('main_category')['discount_percentage'].mean()
plt.figure(figsize=(10, 5))
plt.bar(avg_discount.index, avg_discount.values, color='lightgreen')
plt.xlabel('Product Categories')
plt.ylabel('Average Discount (%)')
plt.title('Average Discount by Product Category')
plt.xticks(rotation=90)
plt.show()
```

```python
# Histogram: Distribution of product ratings

plt.figure(figsize=(8, 5))
plt.hist(df['rating'], bins=10, color='skyblue', edgecolor='black')
plt.xlabel('Product Rating')
plt.ylabel('Frequency')
plt.title('Distribution of Product Ratings')
plt.show()
```

```python
# Pie Chart: Product distribution by category

labels = df['main_category'].value_counts().index
sizes = df['main_category'].value_counts().values
colors = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue','orange', 'pink', 'lightblue', 'violet', 'lightgreen']
explode = [0.1] + [0] * (len(sizes) - 1)  
plt.figure(figsize=(10, 7))
plt.pie(sizes,explode=explode,labels=None,colors=colors,autopct=lambda p: f'{p:.1f}%' if p >= 2 else '',shadow=True,startangle=90)
plt.legend(labels, title='Product Categories',loc='center left', bbox_to_anchor=(1, 0.5))
plt.title('Product Distribution by Category')
plt.show()
```
